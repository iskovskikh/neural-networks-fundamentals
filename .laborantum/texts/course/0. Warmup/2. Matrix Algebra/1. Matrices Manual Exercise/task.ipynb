# NumPy Dot-Product Operator

In NumPy, there is a dot-product operator. To calculate
the dot product of matrices `A` and `B`, run: 
`A @ B`, where `@` is an operator, like `+` or
`*`.

## Task

You are given two matrices `A` and `B` as `numpy.ndarray` objects. The task is to calculate the product
`A @ B` using the standard NumPy dot-product operator: `@`. 
Assume the matrix shapes are compatible for the dot product.

## Interface

**Input**: `A, B: numpy.ndarrays`

**Output**: `C: numpy.ndarray`

```python
def dot_product(A, B):
    res = A
    ## YOUR CODE HERE
    res = A @ B
    return res
```

```python
answer['task5'] = []
for one_input in inputs['task5']:
    answer['task5'].append(dot_product(**one_input))
```
